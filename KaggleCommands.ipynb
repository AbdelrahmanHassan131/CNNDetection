```python
import zipfile
import os

input_dir = "/kaggle/input/diffganmi"
output_dir = "/kaggle/working"

# Make sure output directory exists
os.makedirs(output_dir, exist_ok=True)

# List all parts in order
parts = [
    "dataset_part1.lgnotzip",
    "dataset_part2.lgnotzip",
    "dataset_part3.lgnotzip",
    "dataset_part4.lgnotzip",
    "dataset_part5.lgnotzip",
    "dataset_part6.lgnotzip",
    "dataset_part7.lgnotzip",
    "dataset_part8.lgnotzip",
    # add more parts if you have
]

for part in parts:
    zip_path = os.path.join(input_dir, part)
    print(f"🔹 Extracting {zip_path} ...")
    
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(output_dir)
    
    print(f"✅ Finished extracting {part}")

print(f"🎉 All parts extracted into {output_dir}")
%cd /kaggle/working
!git clone --branch ThesisModelKaggle https://github.com/AbdelrahmanHassan131/CNNDetection.git
%cd /kaggle/working/CNNDetection
!python train_FFT_FreqNet_512.py --dataroot ../dataset --train_split train --val_split val --batch_size 32 --lr 0.0001 --niter 4 --num_threads 4 --name fft_freqnet_512_kaggle
```
